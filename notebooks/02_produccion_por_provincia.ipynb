```python
import pandas as pd

df = pd.read_csv('../data/produccion_energia.csv')
#df.info()                                  # tipos de dato y nulos

df_limpio = df.dropna(subset=['prod_pet'])  # saco filas sin dato clave

# Filtrar solo un año
df_2017 = df_limpio[df_limpio['anio'] == 2017]

# Producción total de petróleo por provincia
resumen = df_2017.groupby('cuenca').agg({'prod_pet':'sum'})
print(resumen)
```

```python
import sqlite3

# Creo una base SQLite en memoria y cargo el dataframe como tabla
conn = sqlite3.connect(':memory:')
df.to_sql('pozos', conn, index=False)

# Ahora sí, SQL de verdad:
query = """
SELECT provincia, SUM(prod_pet) AS total_petroleo
FROM pozos
WHERE anio = 2017
GROUP BY provincia
ORDER BY total_petroleo DESC
"""
resultado_sql = pd.read_sql(query, conn)
print(resultado_sql)
```
