```python
def calcular_produccion_acumulada(dias, tasa_declive):
    # tasa_declive representa cuánto cae la producción por día
    total = dias * (1 - tasa_declive)
    return total
 
# Ahora podés "llamar" a la función cuantas veces quieras:
resultado1 = calcular_produccion_acumulada(30, 0.08)
resultado2 = calcular_produccion_acumulada(60, 0.12)
 
print( "Produccion acumulada: ",resultado1)
print( "Produccion acumulada: ",resultado2)
```

```python
produccion_semana = [820, 795, 830, 780, 810]
 
def calcular_promedio(lista_produccion):
    return sum(lista_produccion) / len(lista_produccion)
 
promedio = calcular_promedio(produccion_semana)
print("Promedio de la semana:", promedio)
 
for dia, produccion in enumerate(produccion_semana):
    if produccion > promedio:
        print(f"Día {dia}: {produccion} m³ → por ENCIMA del promedio")
    else:
        print(f"Día {dia}: {produccion} m³ → por DEBAJO del promedio")
```

```python

# Producción diaria del pozo durante 5 días (m³)
produccion = [120, 135, 110, 150, 125]


# Función para calcular el promedio
def calcular_promedio(lista):
    promedio = sum(lista) / len(lista)
    return promedio


# Calculamos el promedio
promedio = calcular_promedio(produccion)

print(f"El promedio de producción es: {promedio:.2f} m³")


# Recorremos la lista y comparamos cada día con el promedio
for dia, produccion_dia in enumerate(produccion, start=1):

    if produccion_dia > promedio:
        print(f"Día {dia}: {produccion_dia} m³ → estuvo por encima del promedio.")
    else:
        print(f"Día {dia}: {produccion_dia} m³ → no estuvo por encima del promedio.")

```
